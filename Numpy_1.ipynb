```python
# What are the difference between list and numpy array
# 1) In list element wise operation is not possible , but in an numpy array element wise operation is possible.
# 2) Numpy.Array is Homogenrous
# 3) We can convert list to array and also we can convert array to list. But when we convert the array to
#    list, we loose the dimensions.
```

```python
sales = [4000,5000,8000,7000]      # Quarterly Sales
```

```python
sales / 3                  # Element wise operation not possible in list
```

```python
sales
```

```python
sales_arr = np.array(sales)
```

```python
sales_arr
```

```python
sales_arr / 3            # Element wise operation is possible.
```

```python
sales
```

```python
averages = []
for i in sales:
    averages.append(i/3)
print(averages)
```

```python
# Create a 1D Array
l = [1,2,5,6,4,5]
arr1 = np.array(l)
print("Dimension of the array :",arr1.ndim)
print(arr1)
```

```python
# Create a 2D Array
l = [[1,2,5,6,4,5]]
arr2 = np.array(l)
print("Dimension of the array :",arr2.ndim)
print(arr2)
```

```python
# Create a 3D Array
l = [[[1,2,5,6,4,5]]]
arr3 = np.array(l)
print("Dimension of the array :",arr3.ndim)
print(arr3)
```
